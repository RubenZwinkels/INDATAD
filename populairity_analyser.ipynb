# Populairity analyser
## In deze file wordt de populariteit van een video gebaseerd op basis van de bijbehorende gegevens.

### Ik ga beginnen met een aantal functies om populariteits factoren mee te berekenen

De eerste factor is de engagement. Dit wordt bepaald door de views en likes voor elkaar te delen.
Zo weet je of veel kijkers de video ook echt liken.

```python
def calculate_engagement(current_views, current_likes):
    engagement_rate = current_likes/current_views
    return engagement_rate
```

Dit is natuurlijk mogelijk door puur te kijken naar toenemende views, maar ik kies ervoor om het om te rekenen naar een groeipercentage.
