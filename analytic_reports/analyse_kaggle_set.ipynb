# beschrijvende analyse op de statistishe data (van kaggle)

```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
from collections import Counter

df = pd.read_csv("../popularity/Kaggle_TED_video_metadata_balanced.csv")
df.info()
```

```python
fig = px.histogram(df, x='views', nbins=20, title='verdeling van views')
fig.show()
```

```python
df['published_at'] = pd.to_datetime(df['published_at'])

fig = px.scatter(df, x='published_at', y='views', title='Views over Tijd')
fig.show()
```

```python
# Scatter plot voor views en likes
fig = px.scatter(df, x='views', y='likes', title='Relatie tussen Views en Likes')
fig.show()
```

```python
fig = px.violin(df, x='category_id', y='likes', title='Verdeling van Likes per Categorie', box=True, points="all")
fig.show()
```

```python
# Correlatie matrix berekenen
corr_matrix = df[['views', 'likes', 'comment_count']].corr()

fig = go.Figure(data=go.Heatmap(
        z=corr_matrix.values,
        x=['views', 'likes', 'comment_count'],
        y=['views', 'likes', 'comment_count'],
        colorscale='Viridis'))

fig.update_layout(title='Correlatie Heatmap van Views, Likes, Dislikes en Commentaren')
fig.show()
```

##### In deze heatmap is te concluderen dat er erg hoge correlatie is tussen alle factoren.

```python
# Duur omzetten naar minuten (veronderstel dat de duur in de vorm van "HH:MM:SS" is)
df['duration_seconds'] = pd.to_timedelta(df['duration']).dt.total_seconds() / 60  # Omzetten naar minuten

# Scatterplot voor Duur vs. Views
fig = px.scatter(df, x='duration_seconds', y='views', title='Duur van Video vs. Aantal Views', labels={'duration_seconds': 'Duur (minuten)'})
fig.show()
```
